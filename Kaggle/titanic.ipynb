### IMPROVED RF 3

```python
train = train_data.copy()
test = test_data.copy()

train.head()
```

```python
import numpy as np

train['Deck'] = train['Cabin'].apply(lambda x: str(x)[0] if pd.notnull(x) else 'U')
test['Deck'] = test['Cabin'].apply(lambda x: str(x)[0] if pd.notnull(x) else 'U')
train.head()
```

```python
train['Deck'].unique()
```

```python
from sklearn.model_selection import train_test_split

# Deck T is only one specific cabin, belonging to a first class passenger. We can replace it with A, which is the most common first class cabin.
train['Deck'] = train['Deck'].replace('T', 'A')

y = train.Survived
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Deck', 'Embarked']
X = pd.get_dummies(train[features])

test = pd.get_dummies(test[features])

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)
```

```python
X.head()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

model = RandomForestClassifier(n_estimators=1000, max_depth=6, min_samples_split=6, min_samples_leaf=2, random_state=1)
model.fit(train_X, train_y)
preds = model.predict(val_X)
print(accuracy_score(val_y, preds))
```

```python
model = RandomForestClassifier(n_estimators=1000, max_depth=6, min_samples_split=6, min_samples_leaf=2, random_state=1)
model.fit(X, y)
predictions = model.predict(test)

output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")
```

### Decision Tree Class w TTS and Best Model - accuracy score (leaf nodes)

```python
train = train_data.copy()

train.head()
```

```python
y = train.Survived

features = ['Pclass', 'Fare', 'Age']
X = train[features]

X.head()
```

```python
from sklearn.model_selection import train_test_split

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier

def get_accuracy(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeClassifier(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    accuracy = accuracy_score(val_y, preds_val)
    return(accuracy)

# mean absolute error - regression, accuracy score - classification
```

```python
# compare MAE with differing values of max_leaf_nodes
for max_leaf_nodes in [5, 50, 500, 5000]:
    my_acc = get_accuracy(max_leaf_nodes, train_X, val_X, train_y, val_y)
    print("Max leaf nodes: %d  \t\t Accuracy:  %.4f" %(max_leaf_nodes, my_acc))
```

```python
best_model = DecisionTreeClassifier(max_leaf_nodes=50, random_state=0)
best_model.fit(X, y)

predictions = best_model.predict(test_data[features])

output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")
```
